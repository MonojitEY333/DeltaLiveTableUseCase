```python
import dlt 
#creating streaming live table for sales

sales_table_schema ="""
order_id INT ,customer_id INT,transaction_id INT ,product_id INT ,quantity INT,discount_amount DECIMAL(10,2) ,total_amount int ,order_date date """
@dlt.table(
    comment = "sales streaming bronze table",
    table_properties = {
        "layer" : "bronze"
    },
    name = "sales_bronze"
)

def sales_bronze():
  return (
    spark.readStream.csv("/Volumes/ecommerce/dbo/landing/raw/sales.csv", schema= sales_table_schema, header=True))


```
